```python
import os
from dotenv import load_dotenv
import spotipy
from spotipy.oauth2 import SpotifyOAuth

# 1. Load the variables from the .env file
load_dotenv()

# 2. Access them using os.getenv
# This looks for the names you wrote in the .env file
client_id = os.getenv('SPOTIPY_CLIENT_ID')
client_secret = os.getenv('SPOTIPY_CLIENT_SECRET')
redirect_uri = os.getenv('SPOTIPY_REDIRECT_URI')

# 3. Initialize Spotipy
sp = spotipy.Spotify(auth_manager=SpotifyOAuth(
    client_id=client_id,
    client_secret=client_secret,
    redirect_uri=redirect_uri,
    scope="user-library-read"
))

# Test it
user = sp.current_user()
print(f"Successfully authenticated as: {user['display_name']}")
```

```python
import pandas as pd

# The official "Today's Top Hits" playlist
playlist_id = "37i9dQZF1DXcBWIGoYBM5M"

results = sp.playlist_items(playlist_id)['items']
track_uris = [x["track"]["uri"] for x in results]

# Create a list of track details
track_details = []
for uri in track_uris:
    track = sp.track(uri)
    track_details.append({
        'name': track['name'],
        'artist': track['artists'][0]['name'],
        'popularity': track['popularity'],
        'duration_ms': track['duration_ms']
    })

# Convert the list of dictionaries to a pandas DataFrame
df_playlist = pd.DataFrame(track_details)

print(df_playlist.head())
```
