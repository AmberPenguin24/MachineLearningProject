```python
import os
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
from dotenv import load_dotenv
import pandas as pd

# This command tells Python to look at your .env file and 
# "obtain" the credentials so they can be used for the token.
load_dotenv()

auth_manager = SpotifyClientCredentials(
    client_id=os.getenv('SPOTIPY_CLIENT_ID'),
    client_secret=os.getenv('SPOTIPY_CLIENT_SECRET')
)

# This line is where the 'Access Token' is actually requested and stored
sp = spotipy.Spotify(client_credentials_manager=auth_manager)

print("Authentication successful - Access Token obtained!")
```

```python
# 1. Provide the ID from the documentation example
playlist_id = '3qQwZvSUjAxIi3BkKOhuMp'

# 2. Use the .playlist() method to perform the GET request
playlist_data = sp.playlist(playlist_id)

print("Playlist data retrieved successfully!")
print(playlist_data)  # Print the first few rows of the playlist data
```
