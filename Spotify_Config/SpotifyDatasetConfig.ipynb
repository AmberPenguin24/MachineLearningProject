```python
import os
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
from spotipy.oauth2 import SpotifyOAuth
from dotenv import load_dotenv
import pandas as pd

load_dotenv(override=True)

load_dotenv()
cid = os.getenv('SPOTIPY_CLIENT_ID')
secret = os.getenv('SPOTIPY_CLIENT_SECRET')
redirect = os.getenv('SPOTIPY_REDIRECT_URI')

scope = "user-library-read playlist-modify-public"

sp = spotipy.Spotify(auth_manager=SpotifyOAuth(client_id=cid, client_secret=secret, redirect_uri=redirect, scope=scope))

# The first time this runs, a browser window will open for user login/consent
user = sp.current_user()
print(f"Authenticated as: {user['display_name']}")

# Example: Search for an artist
results = sp.search(q='artist:Weezer', type='artist')
print(results['artists']['items'][0]['name'])

try:
    test_track = sp.search(q='track:Stay', type='track', limit=1)
    print("Authentication refreshed and working!")
except Exception as e:
    print(f"Auth failed: {e}")
```

```python
def get_playlist_from_user():
    playlist = input("Enter the Spotify playlist URL that would want songs from: ")

    vibe = input("What vibe are you looking for? Choose from calm, energetic, gloomy, or happy: ")

    playlist_id = playlist.split("/")[-1].split("?")[0] # Extract the playlist ID from the URL

    print(f"Playlist ID extracted: {playlist_id}")

    while(vibe not in ['calm', 'energetic', 'gloomy', 'happy']):
        print("Invalid vibe. Please choose from calm, energetic, gloomy, or happy.")
        vibe = input("What vibe are you looking for? Choose from calm, energetic, gloomy, or happy: ")

    return playlist_id #, vibe

def get_values_from_playlist(sp, playlist_id):
    # This method is better because it goes straight to the track list
    results = sp.playlist_items(playlist_id)
    items = results['items']

    playlist_song_ids = []

    for item in items:
        track_id = item['item']['id']
        track_name = item['item']['name']
        playlist_song_ids.append([track_name, track_id])

    return playlist_song_ids      



pid = get_playlist_from_user()
get_values_from_playlist(sp, pid)
```
